# Retrieval RAG Metrics Lab
Run examples comparing TF-IDF, BM25, vector search, hybrid search, reranking, Recall@K, MRR, and nDCG.

```python
from app.search_engines import TfidfSearch, BM25Search, VectorSearch, HybridSearch
from app.evaluate import evaluate
query = 'ADHD medication side effects'
engines = {
    'tfidf': TfidfSearch(),
    'bm25': BM25Search(),
    'vector': VectorSearch(),
    'hybrid': HybridSearch(),
    'hybrid_rerank': HybridSearch(),
}
```

```python
for name, engine in engines.items():
    print('\n', name.upper())
    results = engine.search(query, k=5, rerank=(name == 'hybrid_rerank')) if name.startswith('hybrid') else engine.search(query, k=5)
    for r in results:
        print(r['rank'], r['id'], r['title'], round(r['score'], 3), round(r.get('rerank_score', 0), 3))
```
