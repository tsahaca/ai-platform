```python
import sys\nfrom pathlib import Path\nsys.path.append(str(Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()))\nimport matplotlib.pyplot as plt\nfrom sklearn.manifold import TSNE\nfrom src.data import DOCUMENTS\nfrom src.embeddings import embed_texts\n
```

```python
texts = [d['title'] + ' ' + d['text'] for d in DOCUMENTS]\nlabels = [d['title'] for d in DOCUMENTS]\nembeddings = embed_texts(texts, normalize=True)\nembeddings.shape\n
```

```python
reduced = TSNE(n_components=2, random_state=42, perplexity=5).fit_transform(embeddings)\nplt.figure(figsize=(10,7))\nfor i, label in enumerate(labels):\n    plt.scatter(reduced[i,0], reduced[i,1])\n    plt.text(reduced[i,0], reduced[i,1], label)\nplt.grid(True)\nplt.show()\n
```
